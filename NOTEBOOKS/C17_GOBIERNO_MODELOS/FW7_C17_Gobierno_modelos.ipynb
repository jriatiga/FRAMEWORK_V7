# **FRAMEWORK V7 DATASET MAESTRO - GOBIERNO DE MODELOS**


### 17.1 Objetivo del Gobierno del Modelo

El Gobierno del Modelo tiene como propósito establecer los mecanismos de control, trazabilidad, documentación, seguimiento y gestión del ciclo de vida de los modelos de Machine Learning desarrollados dentro del Framework V7 para la gestión hídrica.

Esta capa de gobierno busca garantizar que los modelos desarrollados puedan ser identificados, reproducidos, evaluados y utilizados de manera controlada, manteniendo evidencia sobre los datos empleados, las variables predictoras, la configuración experimental, la arquitectura, los artefactos generados y las condiciones bajo las cuales los modelos pueden ser utilizados.

Para este propósito se toma como referencia el experimento **Exp01-V3**, correspondiente a la predicción del **Índice de Riesgo de la Calidad del Agua (IRCA)** mediante una arquitectura **LSTM**, como modelo que alcanzó una versión utilizable después del proceso de diagnóstico y ajuste realizado durante el desarrollo del Framework.

El Gobierno del Modelo no modifica el proceso de preparación, transformación, entrenamiento o evaluación previamente desarrollado en los notebooks del Framework. Su función es establecer una capa de control sobre los resultados generados por dicho proceso, permitiendo documentar y administrar el modelo una vez construido.

En este sentido, C17 integra las evidencias producidas durante las etapas anteriores del Framework, entre ellas:

- configuración de los experimentos;
- selección de variables predictoras;
- transformación de los datos;
- construcción de secuencias temporales;
- generación de tensores;
- entrenamiento del modelo;
- evaluación del desempeño;
- diagnóstico del modelo;
- recomendaciones de mejora;
- bitácoras de ejecución;
- metadata de los tensores;
- archivo del modelo entrenado;
- Model Card y demás artefactos asociados.

El objetivo final es establecer un esquema de gobierno que permita responder, para cada modelo, **qué modelo fue construido, con qué datos y variables, bajo qué configuración, para qué propósito, qué evidencia respalda su utilización, cuáles son sus limitaciones y bajo qué condiciones debe ser actualizado o retirado**.

Para el experimento **Exp01-V3**, la trazabilidad se establece a partir del identificador del experimento y de los artefactos almacenados en el repositorio del Framework. De esta manera, el modelo físico no se considera un elemento aislado, sino parte de un conjunto de evidencias técnicas que permiten reconstruir y auditar su ciclo de vida.

Por tanto, el Gobierno del Modelo establece como principio que **cada modelo desplegado o utilizado como referencia debe poder relacionarse con sus datos, transformaciones, configuración, resultados, diagnósticos y documentación correspondientes**.

### 17.4 Propósito, uso previsto y uso no previsto

El Gobierno del Modelo contempla los modelos que han sido efectivamente desarrollados dentro del Framework V7 y que cuentan con artefactos de entrenamiento, evaluación, diagnóstico y documentación. Para el estado actual del proyecto se consideran como modelos de referencia **Exp01-V3** y **Exp04**, correspondientes a problemas predictivos diferentes dentro del dominio de Gestión Hídrica.

| Modelo | Variable objetivo | Tipo de problema | Propósito |
|---|---|---|---|
| Exp01-V3 | `irca` | Clasificación | Predicción del riesgo sanitario asociado a la calidad del agua |
| Exp04 | `VolumenUtilDiarioMasa` | Regresión | Predicción del volumen útil disponible del sistema |

#### 17.4.1 Exp01-V3 – Predicción del IRCA

El propósito de Exp01-V3 es apoyar el análisis predictivo del riesgo sanitario asociado a la calidad del agua mediante la estimación de la categoría correspondiente al IRCA.

El modelo utiliza una estructura temporal basada en una ventana de 12 observaciones y un horizonte de predicción de 1 periodo. Sus resultados permiten identificar posibles condiciones asociadas con cambios en el comportamiento del indicador.

El modelo está destinado a apoyar el análisis y la toma de decisiones, pero no sustituye las mediciones de calidad del agua, los análisis de laboratorio, los procedimientos regulatorios ni el criterio profesional.

#### 17.4.2 Exp04 – Predicción del volumen útil

El propósito de Exp04 es apoyar el análisis predictivo de la disponibilidad del recurso hídrico mediante la estimación de la variable `VolumenUtilDiarioMasa`.

El modelo busca identificar el comportamiento esperado de esta variable a partir de la información disponible en el conjunto de variables predictoras y de su comportamiento temporal.

Sus resultados pueden utilizarse como apoyo para el análisis de disponibilidad del recurso, identificación de tendencias y evaluación de escenarios de gestión.

Al igual que Exp01-V3, sus resultados constituyen estimaciones del modelo y no deben interpretarse como sustituto de mediciones directas, sistemas de medición operativa o criterios técnicos establecidos para la gestión del recurso.

#### 17.4.3 Usos previstos

Los modelos actualmente considerados dentro del Gobierno del Modelo pueden utilizarse para:

- apoyar el análisis predictivo de variables relacionadas con la gestión hídrica;
- anticipar posibles comportamientos de las variables objetivo;
- complementar el análisis histórico y observacional;
- apoyar la identificación de situaciones que requieran revisión técnica;
- generar información para escenarios de análisis dentro del Framework;
- servir como componentes predictivos de los procesos de Machine Learning desarrollados.

El uso de cada modelo deberá respetar su variable objetivo, conjunto de variables predictoras, estructura temporal y configuración experimental correspondiente.

#### 17.4.4 Usos no previstos

Los modelos no deben utilizarse como mecanismos autónomos para:

- sustituir mediciones directas o análisis de laboratorio;
- emitir decisiones regulatorias sin validación técnica;
- establecer relaciones causales únicamente a partir de las predicciones;
- realizar predicciones utilizando variables que no correspondan con las utilizadas durante su entrenamiento;
- aplicar un modelo a un dominio diferente del definido en su experimento;
- asumir que una predicción mantiene el mismo nivel de confiabilidad ante condiciones que no estén representadas adecuadamente en los datos utilizados para su desarrollo;
- tomar decisiones críticas sin supervisión y validación humana.

En particular, **Exp01-V3 no debe utilizarse para predecir el volumen útil**, así como **Exp04 no debe utilizarse para predecir el IRCA**. Cada modelo mantiene su propósito y dominio de aplicación definidos en el catálogo experimental.

#### 17.4.5 Límites de aplicabilidad

La aplicabilidad de ambos modelos depende de la calidad, representatividad y estructura temporal de los datos utilizados durante su construcción.

Entre las principales restricciones se encuentran:

1. **Representatividad:** condiciones ambientales o hidrológicas no presentes en los datos históricos pueden afectar la capacidad de generalización.

2. **Calidad de las entradas:** valores erróneos, faltantes, cambios en unidades o modificaciones en los mecanismos de captura pueden afectar las predicciones.

3. **Estructura temporal:** ambos experimentos utilizan una ventana temporal de 12 observaciones y un horizonte de 1 periodo. Cambiar esta estructura implica una nueva configuración experimental y requiere evaluación.

4. **Variables predictoras:** cada modelo depende del conjunto de variables definido para su experimento. La modificación de estas variables debe ser registrada y evaluada.

5. **Dominio de aplicación:** cada modelo debe utilizarse exclusivamente para la variable objetivo y el problema para el cual fue desarrollado.

6. **Generalización:** el desempeño observado durante la evaluación no garantiza el mismo comportamiento ante periodos, condiciones o poblaciones significativamente diferentes de aquellas representadas en los datos utilizados para el desarrollo.

Por lo anterior, cualquier modificación sustancial del contexto de aplicación, variable objetivo, variables predictoras, estructura temporal o configuración del modelo deberá ser considerada dentro del proceso de control de cambios y Gobierno del Modelo.

#### 17.4.6 Responsabilidad sobre las decisiones

Los modelos desarrollados en el Framework V7 constituyen herramientas de apoyo a la decisión y no transfieren la responsabilidad de las decisiones técnicas o administrativas al sistema predictivo.

Las predicciones generadas por Exp01-V3 y Exp04 deberán ser interpretadas conjuntamente con información observacional, conocimiento del dominio y criterios técnicos antes de adoptar decisiones con consecuencias operativas, ambientales o sanitarias.

#### 17.5.3 Separación entre variables predictoras y objetivo

El Framework establece como principio de gobernanza la separación explícita entre las variables predictoras y la variable objetivo.

Esta separación permite reducir el riesgo de fuga de información y mantener la trazabilidad de las variables que intervienen en cada experimento.

La selección de variables no se realiza directamente dentro del entrenamiento del modelo, sino que se deriva de los artefactos y configuraciones establecidos durante las etapas anteriores del Framework.

Por tanto, cualquier incorporación, eliminación o modificación de una variable predictora deberá quedar registrada como un cambio en la configuración del experimento y deberá generar una nueva versión o ejecución controlada del modelo cuando dicha modificación pueda afectar su comportamiento.

#### 17.5.4 Transformación de los datos

Las transformaciones aplicadas a los datos forman parte del modelo gobernado y deben conservarse junto con los demás artefactos del experimento.

En particular, el objeto de transformación utilizado durante la preparación de los datos debe permanecer disponible para garantizar que futuras inferencias utilicen una transformación consistente con aquella empleada durante el entrenamiento.

De esta forma, el modelo no se considera únicamente como el archivo `.keras`, sino como un conjunto compuesto por:

**datos → transformación → secuencias/tensores → modelo → predicción → diagnóstico.**

Esta consideración resulta especialmente importante para la futura integración del modelo en el componente de inferencia del Framework y evita utilizar durante producción una transformación diferente a la utilizada durante el entrenamiento.

#### 17.5.5 Control de cambios sobre las variables

Cualquier cambio en:

- la variable objetivo;
- las variables predictoras;
- el orden de las variables;
- la ventana temporal;
- el horizonte de predicción;
- el método de transformación;
- o la estructura de los datos de entrada,

deberá ser registrado y evaluado antes de considerar el modelo resultante como equivalente al modelo actualmente gobernado.

Cuando el cambio modifique las características fundamentales del experimento, deberá identificarse como una nueva versión del modelo y conservarse la versión anterior como evidencia histórica.

Este mecanismo permite mantener la trazabilidad entre el modelo utilizado, los datos que recibió y las decisiones metodológicas que determinaron su construcción.

### 17.6 Trazabilidad del ciclo de vida del modelo

El Gobierno del Modelo del Framework V7 se establece como una capa transversal de control sobre los procesos de preparación de datos, Machine Learning, modelado, evaluación e interpretación desarrollados previamente. En consecuencia, C17 no reemplaza ni repite los procesos metodológicos desarrollados en los capítulos anteriores, sino que registra, relaciona y controla los resultados generados por dichos procesos.

El ciclo de vida del modelo se establece mediante la relación entre las diferentes etapas del Framework y los artefactos producidos en cada una de ellas.

#### 17.6.1 Flujo general de trazabilidad

La trazabilidad del modelo se establece mediante la siguiente secuencia:

**Dataset Maestro → Preparación Machine Learning → Diseño Experimental → Transformación de datos → Modelado → Evaluación → Interpretación de Resultados → Gobierno del Modelo → Inferencia/Aplicación**

Esta secuencia permite mantener la relación entre los datos de origen, las decisiones metodológicas, los modelos construidos, sus resultados y las decisiones posteriores relacionadas con su utilización y control.

| Etapa | Componente | Función dentro del ciclo de vida | Evidencia principal |
|---|---|---|---|
| Datos de origen | **C09** | Consolidación del Dataset Maestro | Dataset Maestro |
| Preparación ML | **C12** | Preparación y selección de variables para Machine Learning | Dataset ML, variables seleccionadas |
| Diseño experimental | **C13** | Definición de experimentos y configuración | Catálogo y configuración experimental |
| Transformación | **C13** | Transformación y construcción de secuencias temporales | Dataset transformado, scaler, secuencias y metadata |
| Modelado | **C14** | Construcción y entrenamiento de los modelos | Tensores, registros y modelos `.keras` |
| Evaluación | **C15** | Medición del desempeño de los modelos | Métricas, predicciones y resultados de evaluación |
| Interpretación | **C16** | Análisis e interpretación técnica de los resultados | Interpretaciones, conclusiones y análisis de resultados |
| Gobierno | **C17** | Identificación, versionamiento, trazabilidad y control | Registros de gobierno y documentación |
| Aplicación | **Framework Streamlit** | Visualización e interacción con los resultados y modelos | Aplicación e interfaz de inferencia |

#### 17.6.2 Función de C15 y C16 dentro de la trazabilidad

El Framework diferencia explícitamente entre la **evaluación cuantitativa del modelo** y la **interpretación de sus resultados**.

El componente C15 se concentra en determinar el comportamiento del modelo mediante métricas, predicciones y procedimientos de evaluación.

El componente C16 toma dichos resultados y los interpreta dentro del contexto del problema de gestión hídrica, permitiendo establecer conclusiones sobre el comportamiento predictivo, posibles limitaciones y utilidad de los resultados.

Por tanto:

**C15 responde principalmente a:**

> ¿Cómo se comportó el modelo?

Mientras que:

**C16 responde principalmente a:**

> ¿Qué significan esos resultados dentro del problema de gestión hídrica?

Finalmente, **C17 responde a:**

> ¿Cómo se identifica, controla, documenta, versiona y supervisa el modelo a partir de la evidencia generada?

Esta separación evita confundir una métrica de desempeño con una conclusión de carácter científico o de dominio.

#### 17.6.5 Relación entre evaluación, interpretación y gobierno

- el desempeño observado del modelo;

**Evaluación → Interpretación → Evidencia para Gobierno**

#### 17.6.6 Relación entre artefactos y modelo

El Gobierno del Modelo considera que el archivo `.keras` por sí solo no constituye una representación suficiente del modelo gobernado.

Para garantizar reproducibilidad y trazabilidad, el modelo debe mantenerse asociado a los elementos que permiten identificar las condiciones bajo las cuales fue generado.

En consecuencia, la unidad de gobierno está constituida por:

**Modelo + configuración + transformación + datos de entrada + tensores/secuencias + metadata + evaluación + interpretación + diagnóstico + recomendaciones + bitácora.**

Esta relación permite determinar no solamente qué modelo fue utilizado, sino también:

- con qué datos fue construido;
- qué variables fueron utilizadas;
- qué transformación se aplicó;
- qué estructura temporal se utilizó;
- qué configuración experimental lo originó;
- qué resultados obtuvo;
- qué interpretación se realizó;
- qué problemas fueron identificados;
- qué recomendaciones fueron generadas;
- y qué versión del modelo fue registrada.

#### 17.8.3 Criterios de aceptación

El Gobierno del Modelo establece que una métrica por sí sola no constituye evidencia suficiente para declarar que un modelo es adecuado para su utilización.

La aceptación de un modelo deberá considerar conjuntamente:

1. **Desempeño predictivo:** comportamiento de las métricas correspondientes al tipo de problema.
2. **Consistencia temporal:** comportamiento del modelo sobre datos respetando el orden cronológico.
3. **Ausencia de fuga de información:** separación adecuada entre información utilizada para entrenamiento y evaluación.
4. **Estabilidad:** comportamiento razonablemente consistente ante diferentes segmentos temporales.
5. **Interpretabilidad:** coherencia entre los resultados obtenidos y el contexto de gestión hídrica.
6. **Trazabilidad:** existencia de configuración, metadata, registros y artefactos asociados.
7. **Limitaciones conocidas:** identificación de condiciones bajo las cuales el modelo no debe utilizarse.
8. **Reproducibilidad:** disponibilidad de los artefactos necesarios para reconstruir el proceso.

#### 17.8.4 Validación de acuerdo con la naturaleza temporal de los datos

Los datos utilizados por el Framework presentan una estructura temporal. Por esta razón, los mecanismos tradicionales de validación aleatoria, como un `K-Fold` convencional, no constituyen necesariamente la estrategia más apropiada.

Para una evaluación adicional orientada a fortalecer el rigor estadístico del modelo, se establece como estrategia recomendada la utilización de:

**Time Series Cross-Validation / Walk-Forward Validation.**

Esta estrategia mantiene el orden temporal de las observaciones y evita utilizar información futura para evaluar periodos anteriores.

De manera conceptual, el procedimiento puede representarse como:

**Entrenamiento inicial → validación futura → ampliación de entrenamiento → nueva validación futura → repetición**

Este procedimiento permite obtener varias estimaciones del desempeño del modelo en diferentes segmentos temporales.

#### 17.8.5 Diferenciación entre resultados originales y validación adicional

Es importante diferenciar entre:

**Evaluación original del experimento**

y

**Validación temporal adicional para fortalecimiento metodológico.**

Los resultados registrados de Exp01-V3 y Exp04 corresponden a los procedimientos de evaluación implementados durante el desarrollo original del Framework y deben conservarse sin modificación.

La incorporación posterior de una estrategia `TimeSeriesSplit` o `Walk-Forward Validation` no reemplaza dichos resultados ni los presenta como si hubieran sido obtenidos originalmente mediante validación cruzada.

En caso de ejecutarse esta evaluación adicional, sus resultados deberán identificarse como una nueva evidencia de validación asociada a la versión del modelo correspondiente.

#### 17.8.6 Intervalos de confianza y variabilidad

La guía institucional establece la necesidad de reportar la variabilidad de los resultados y, cuando corresponda, intervalos de confianza.

En consecuencia, una validación temporal futura podrá utilizar los resultados obtenidos en cada partición temporal para calcular:

- media de la métrica;
- desviación estándar;
- rango de resultados;
- intervalos de confianza;
- estabilidad entre periodos.

La presentación de estos valores permitirá complementar la evaluación puntual de los modelos con una medida de incertidumbre sobre su desempeño.

Para clasificación podrán analizarse métricas como Accuracy, Precision, Recall y F1, según la distribución de las clases y los objetivos del experimento.

Para regresión podrán analizarse métricas como RMSE y MAE, entre otras que resulten pertinentes.

#### 17.8.7 Control de sobreajuste

El control de sobreajuste constituye otro elemento de evaluación del modelo.

En los modelos LSTM utilizados por el Framework, la evaluación debe considerar la diferencia entre el comportamiento observado durante entrenamiento y validación, así como las curvas de pérdida y las métricas registradas durante el proceso de entrenamiento.

El análisis de sobreajuste debe realizarse conjuntamente con:

- curvas de entrenamiento;
- curvas de validación;
- comportamiento de la pérdida;
- comportamiento de la métrica;
- resultados sobre datos no utilizados durante el entrenamiento;
- diagnóstico del modelo.

La presencia de una arquitectura recurrente y de un mecanismo de regularización no elimina por sí misma el riesgo de sobreajuste; este debe evaluarse mediante evidencia empírica.

#### 17.8.8 Fuga de información

El Gobierno del Modelo establece como criterio obligatorio la revisión de posibles fugas de información.

En particular, deberá verificarse que:

- la variable objetivo no sea utilizada directamente como entrada cuando no corresponda;
- la transformación de datos sea consistente con el procedimiento utilizado durante entrenamiento;
- la información futura no sea utilizada para predecir periodos anteriores;
- los procesos de escalamiento sean aplicados respetando la separación temporal;
- las variables utilizadas en cada experimento correspondan a las registradas en su metadata.

Este control resulta especialmente relevante en modelos de series temporales, debido a que una partición incorrecta de los datos puede producir estimaciones de desempeño artificialmente optimistas.

### 17.9 Control de riesgos, limitaciones y uso seguro del modelo

El Gobierno del Modelo incorpora un mecanismo de identificación y control de riesgos orientado a evitar usos incorrectos de los modelos desarrollados en el Framework V7.

Los riesgos considerados se derivan de la naturaleza de los datos, la estructura temporal, las características de los modelos utilizados y los resultados obtenidos durante los procesos de diagnóstico e interpretación.

C17 no reemplaza los diagnósticos realizados durante C15 ni la interpretación desarrollada en C16. Estos componentes constituyen las fuentes de evidencia utilizadas para establecer los controles de gobierno.

#### 17.9.1 Principales riesgos identificados

Los principales riesgos asociados al ciclo de vida de los modelos son:

| Riesgo | Descripción | Control de Gobierno |
|---|---|---|
| Fuga de información | Uso directo o indirecto de información que no debería estar disponible al momento de la predicción | Revisión de variables, transformación y separación temporal |
| Sobreajuste | El modelo aprende características específicas de los datos de entrenamiento que no se generalizan | Evaluación sobre datos no utilizados y análisis de curvas de entrenamiento |
| Cambio de distribución | Las características de los datos pueden cambiar con el tiempo | Monitoreo periódico y evaluación de drift |
| Datos fuera del dominio | Ingreso de valores que se encuentran fuera de las condiciones observadas durante el entrenamiento | Validación de rangos y advertencias en inferencia |
| Datos faltantes | Ausencia de variables necesarias para generar una predicción | Validación de entradas y manejo de excepciones |
| Cambio metodológico | Modificación de variables, transformaciones o arquitectura sin control de versión | Versionamiento de experimentos y modelos |
| Interpretación incorrecta | Utilización de una predicción como certeza absoluta | Presentación de resultados junto con sus limitaciones |
| Dependencia temporal | El desempeño puede variar entre diferentes periodos | Validación temporal y monitoreo del desempeño |
| Deriva del modelo | Pérdida progresiva de desempeño después del despliegue | Revisión periódica y criterios de reentrenamiento |
| Uso no previsto | Utilización del modelo para decisiones diferentes de aquellas para las que fue desarrollado | Definición explícita del propósito y restricciones de uso |

#### 17.9.2 Riesgo de fuga de información

La fuga de información constituye uno de los riesgos más relevantes para modelos de series temporales.

El Gobierno del Modelo establece que las variables utilizadas para generar una predicción deben corresponder exclusivamente a información disponible en el momento temporal en que dicha predicción sería realizada.

En consecuencia, no deberán incorporarse variables que contengan directa o indirectamente información futura respecto al horizonte de predicción.

También deberá verificarse que los procesos de transformación y escalamiento no utilicen información proveniente de periodos que deberían permanecer fuera del conjunto de entrenamiento.

Este control deberá revisarse especialmente cuando se generen nuevas versiones de los modelos.

#### 17.9.3 Riesgo de sobreajuste

El sobreajuste puede producir modelos con un desempeño elevado durante entrenamiento pero con una capacidad de generalización limitada.

En los modelos LSTM del Framework, este riesgo se controla mediante:

- separación entre entrenamiento y evaluación;
- seguimiento de la pérdida durante el entrenamiento;
- seguimiento de las métricas;
- utilización de mecanismos de regularización cuando corresponda;
- análisis de las curvas de entrenamiento;
- evaluación sobre datos no utilizados para el entrenamiento;
- validación temporal adicional.

Para Exp01-V3, la arquitectura incluye una capa Dropout, constituyendo un mecanismo explícito de regularización.

No obstante, la presencia de Dropout no constituye por sí misma evidencia de ausencia de sobreajuste. El comportamiento debe continuar siendo evaluado mediante los resultados obtenidos.

#### 17.9.4 Cambio de distribución de los datos

Los modelos fueron construidos a partir de datos históricos y, por tanto, existe la posibilidad de que las condiciones futuras difieran de aquellas observadas durante su entrenamiento.

Entre las posibles causas se encuentran:

- cambios climáticos;
- modificaciones en los patrones de precipitación;
- cambios en las condiciones hidrológicas;
- cambios en la calidad del agua;
- modificaciones en las fuentes de información;
- cambios en los procedimientos de medición;
- incorporación de nuevos rangos de valores.

Cuando la distribución de los datos de entrada cambie significativamente, las predicciones pueden perder confiabilidad aunque el modelo físico continúe funcionando correctamente.

Por esta razón, el Gobierno del Modelo incorpora el concepto de **drift** como elemento de monitoreo posterior.

#### 17.9.5 Datos fuera del dominio de entrenamiento

Una predicción generada a partir de valores muy diferentes de aquellos observados durante el entrenamiento debe considerarse de mayor incertidumbre.

La aplicación de inferencia deberá validar, cuando sea técnicamente posible:

- presencia de las variables requeridas;
- tipos de datos;
- valores faltantes;
- rangos razonables;
- estructura temporal;
- consistencia entre las variables;
- compatibilidad con la transformación utilizada durante el entrenamiento.

Cuando una entrada no cumpla las condiciones mínimas, el sistema deberá evitar presentar la predicción como un resultado normal y deberá generar una advertencia o mensaje de validación.

#### 17.9.6 Uso previsto

Los modelos desarrollados dentro del Framework tienen como propósito apoyar el análisis predictivo de variables relacionadas con la gestión hídrica.

En particular:

**Exp01-V3**

Su propósito es apoyar la anticipación del riesgo sanitario asociado al agua mediante la predicción del IRCA a partir de información temporal de variables ambientales e hidrológicas.

**Exp04**

Su propósito es apoyar el análisis predictivo relacionado con el comportamiento de `VolumenUtilDiarioMasa`, contribuyendo al análisis de disponibilidad y gestión del recurso hídrico.

Las predicciones deben entenderse como **herramientas de apoyo a la toma de decisiones**, y no como sustitutos de los procesos técnicos, institucionales o regulatorios relacionados con la gestión del agua.

#### 17.9.7 Uso no previsto

Los modelos no deberán utilizarse para:

- establecer diagnósticos sanitarios definitivos;
- reemplazar mediciones oficiales;
- sustituir análisis de laboratorio;
- establecer decisiones regulatorias de manera automática;
- garantizar condiciones futuras del recurso hídrico;
- realizar predicciones sobre variables diferentes de la variable objetivo para la cual fueron entrenados;
- generar predicciones cuando las variables requeridas no estén disponibles;
- extrapolar indefinidamente fuera de las condiciones observadas durante el entrenamiento;
- utilizar versiones del modelo diferentes de la versión oficialmente registrada;
- interpretar una predicción individual como certeza absoluta.

Cualquier utilización diferente del propósito establecido deberá ser evaluada antes de incorporarse al ciclo de vida gobernado.

#### 17.9.8 Riesgo de interpretación de las predicciones

Una predicción generada por un modelo de Machine Learning representa una estimación basada en patrones aprendidos a partir de datos históricos.

Por esta razón, el resultado debe interpretarse conjuntamente con:

- las condiciones de entrada;
- el periodo temporal;
- el desempeño histórico del modelo;
- las limitaciones conocidas;
- la calidad de los datos;
- y el contexto específico de la decisión.

El sistema deberá evitar presentar una predicción como una certeza determinística.

#### 17.9.9 Control de versiones frente a cambios

Una nueva versión deberá generarse cuando exista una modificación relevante en:

- datos;
- variables predictoras;
- variable objetivo;
- transformación;
- ventana temporal;
- horizonte;
- arquitectura;
- hiperparámetros;
- procedimiento de entrenamiento;
- procedimiento de evaluación;
- o criterios de aceptación.

La nueva versión deberá conservar una relación explícita con la versión anterior, permitiendo reconstruir la evolución del modelo.

El caso de **Exp01-V3** constituye un ejemplo de este principio, debido a que la versión utilizada representa una evolución del experimento inicial como resultado del proceso de diagnóstico y ajuste.

#### 17.9.10 Monitoreo posterior al despliegue

Cuando los modelos sean utilizados mediante una aplicación de inferencia, deberá establecerse un mecanismo de monitoreo que permita identificar cambios en su comportamiento.

Como mínimo, el monitoreo deberá considerar:

- cantidad de predicciones generadas;
- valores de entrada;
- errores de inferencia;
- disponibilidad de las variables;
- distribución de las entradas;
- distribución de las predicciones;
- desempeño cuando posteriormente se disponga del valor observado;
- cambios significativos respecto al comportamiento histórico.

El monitoreo permitirá identificar señales de degradación que puedan justificar una nueva evaluación o reentrenamiento.

#### 17.9.11 Criterios para una nueva evaluación o reentrenamiento

Una nueva evaluación deberá considerarse cuando se detecte alguno de los siguientes eventos:

1. deterioro significativo de las métricas;
2. cambio importante en la distribución de las variables;
3. aparición de nuevos rangos de valores;
4. modificación de las fuentes de datos;
5. cambio en la definición de la variable objetivo;
6. incorporación o eliminación de variables predictoras;
7. modificación de la estructura temporal;
8. aparición de nuevos patrones que no estén representados en los datos históricos.

El reentrenamiento no deberá realizarse únicamente por criterio temporal. Deberá estar respaldado por evidencia que justifique la necesidad de actualizar el modelo.

#### 17.9.12 Matriz de riesgo del modelo

Como mecanismo de gobierno se establece la siguiente clasificación inicial:

| Riesgo | Probabilidad | Impacto | Nivel de control |
|---|---|---|---|
| Fuga de información | Media | Alto | Alto |
| Sobreajuste | Media | Alto | Alto |
| Cambio de distribución | Media | Alto | Alto |
| Datos fuera del dominio | Media | Alto | Alto |
| Datos faltantes | Media | Medio | Medio |
| Drift del modelo | Media | Alto | Alto |
| Uso no previsto | Media | Alto | Alto |
| Interpretación incorrecta | Media | Alto | Alto |
| Modificación sin versionamiento | Baja | Alto | Alto |
| Fallos de inferencia | Baja/Media | Medio | Medio |

Esta matriz constituye una línea base de gobierno y podrá ser actualizada cuando se disponga de evidencia operacional adicional.

#### 17.9.13 Estado actual

Los principales riesgos metodológicos han sido considerados a partir de los procesos de preparación, evaluación, diagnóstico e interpretación desarrollados previamente.

Actualmente se dispone de mecanismos de trazabilidad y documentación para relacionar los modelos con sus artefactos y resultados.

Quedan como actividades de fortalecimiento del Gobierno del Modelo:

- implementación de validación temporal adicional;
- establecimiento de métricas de monitoreo de drift;
- definición de umbrales operativos de alerta;
- incorporación de validaciones de entrada en la interfaz de inferencia;
- definición formal del procedimiento de reentrenamiento;
- integración de estos controles en la aplicación Streamlit.

Estas actividades no modifican los modelos históricos ya generados, sino que complementan su gobierno y establecen las condiciones necesarias para su utilización controlada.
